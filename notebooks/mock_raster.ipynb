```python
import numpy as np
import rasterio
from rasterio.transform import from_bounds #kontekst przestrzenny
```

```python
from matplotlib.colors import LightSource #cieniowanie
import os
```

```python
WIDTH, HEIGHT = 1000, 1000
OUTPUT_DIR = "../app/frontend/data/mocks"
```

```python
os.path.isdir(OUTPUT_DIR)
```

```python
#TODO: CHECK WITH VERTICES COORDINATES
```

```python
BOUNDS_3035 = {
    'WEST': 4899342.9771,
    'EAST': 5000581.4977,
    'SOUTH': 2999266.3481,
    'NORTH': 3100504.8686,
}
```

```python
print("generating mocks for hillshade...")
```

```python
y = np.random.uniform(-3, 5, HEIGHT)
x = np.random.uniform(-3, 5, WIDTH)
X, Y = np.meshgrid(x, y)
#hills 
dem = (np.sin(X * 2) + np.cos(Y * 3) + np.sin(X**2 + Y**2) * 0.5) * 150
#shades
ls = LightSource(azdeg=315, altdeg=45)
hillshade=ls.hillshade(dem, vert_exag=1, dx=100, dy=100)
#converting to lighter file (0-255)
hillshade_8bit = (hillshade*255).astype(np.uint8)

transform = from_bounds(
    BOUNDS_3035['WEST'], BOUNDS_3035['SOUTH'],
    BOUNDS_3035['EAST'], BOUNDS_3035['NORTH'],
    WIDTH, HEIGHT
)
```

```python
#save
with rasterio.open(f'{OUTPUT_DIR}/mock_hillshade.tif',
                  'w',
                  driver='GTiff',
                height=HEIGHT,
                width=WIDTH,
                count=1,
                dtype=hillshade_8bit.dtype,
                crs='EPSG:3035',
                transform=transform,
) as dst:
    dst.write(hillshade_8bit, 1)
```

```python
print("Hillshade generated succesfully")
```
